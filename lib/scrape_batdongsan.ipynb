```python
import asyncio
from playwright.async_api import async_playwright
from bs4 import BeautifulSoup

BASE_URL = "https://batdongsan.com.vn"
START_URL = "https://batdongsan.com.vn/nha-dat-ban-phuong-vinh-hoa-2"
MAX_PAGES = 3  # Adjust as needed

async def scrape_page(page, url):
    """Scrapes a single page and extracts listings."""
    print(f"Scraping: {url}")
    await page.goto(url, timeout=60000, wait_until="networkidle")
    html = await page.content()
    soup = BeautifulSoup(html, "html.parser")

    listings = []
    
    for element in soup.select('div.re__srp-list.js__srp-list.clear-after div[class*="js__card js__card-full-web"]'):
        listing_url = element.select_one("a.js__product-link-for-product-id")
        title = element.select_one("span.pr-title.js__card-title")
        price = element.select_one("span.re__card-config-price.js__card-config-item")
        area = element.select_one("span.re__card-config-area.js__card-config-item")
        price_per_m2 = element.select_one("span.re__card-config-price_per_m2.js__card-config-item")
        location = element.select_one("span.re__card-location span:last-child")
        description = element.select_one("div.re__card-description.js__card-description")
        image = element.select_one("img[data-img]")

        listing_data = {
            "listingURL": BASE_URL + listing_url["href"] if listing_url else None,
            "title": title.get_text(strip=True) if title else None,
            "price": price.get_text(strip=True) if price else None,
            "area": area.get_text(strip=True) if area else None,
            "pricePerM2": price_per_m2.get_text(strip=True) if price_per_m2 else None,
            "location": location.get_text(strip=True) if location else None,
            "description": description.get_text(strip=True) if description else None,
            "image": image["data-img"] if image and "data-img" in image.attrs else None,
        }

        if listing_data["title"] and listing_data["price"]:
            listings.append(listing_data)

    return listings

async def scrape_website():
    """Scrapes multiple pages and collects all listings."""
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox", "--disable-setuid-sandbox"])
        context = await browser.new_context(user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36")
        page = await context.new_page()

        all_listings = []
        
        for page_number in range(1, MAX_PAGES + 1):
            url = START_URL if page_number == 1 else f"{START_URL}/p{page_number}"
            listings = await scrape_page(page, url)
            all_listings.extend(listings)
        
        await browser.close()
        return all_listings

async def main():
    listings = await scrape_website()
    
    print("Extracted Listings:")
    for listing in listings:
        print(listing)

# Run the main function
asyncio.run(main())
```
